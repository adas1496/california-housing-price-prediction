```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import fetch_california_housing
```

```python
housing = fetch_california_housing()
```

```python
print(housing)
```

```python
housing = fetch_california_housing(as_frame=True)
```

```python
df = housing.frame
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
df['rooms_per_person'] = df['AveRooms']/df['AveOccup']
df['bedrooms_ratio'] = df['AveBedrms']/df['AveRooms']
df['population_per_household'] = df['Population']/df['AveOccup']
```

```python
print(df)
```

```python
df.head()
```

```python
corr = df.corr()
```

```python
sns.heatmap(corr, cmap='coolwarm', annot=True, fmt='0.2f', linewidths=0.5, annot_kws={'size':9}, cbar_kws={"shrink":0.8})

plt.title('Feature Correlation', fontsize=16)
plt.xticks(rotation=45, ha='right', fontsize=16)
plt.yticks(fontsize=10)
plt.tight_layout
plt.show()
```

```python
X = df.drop('MedHouseVal', axis=1)
y = df['MedHouseVal']
```

```python
print(X)
print(y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
print(X_train)
print(X_test)
```

```python
print(y_train)
print(y_test)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train_scaled)
print(X_test_scaled)
```

```python
!pip install xgboost
```

```python
from xgboost import XGBRegressor
```

```python
xgb_model = XGBRegressor(
    random_state = 42,
    n_jobs=-1
)
```

```python
xgb_model.fit(X_train_scaled, y_train)
```

```python
y_pred = xgb_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("RMSE:",rmse)
print("R2 SCORE:", r2)
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
param_grid = {
    'n_estimators': [100, 200, 300, 400],
    'max_depth': [3, 5, 7, 9],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0],
    'min_child_weight': [1, 3, 5]
}
```

```python
xgb = XGBRegressor(random_state=42)
```

```python
xgb_random = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_grid,
    n_iter=20,
    cv=3,
    verbose=2,
    random_state=42,
    n_jobs=-1
)
```

```python
xgb_random.fit(X_train_scaled, y_train)
```

```python
xgb_random.fit(X_train_scaled, y_train)
```

```python
xgb_random.best_params_
```

```python
xgb_random.best_score_
```

```python
final_xgb = XGBRegressor(
    n_estimators=400,
    max_depth=7,
    learning_rate=0.1,
    subsample=0.9,
    colsample_bytree=0.8,
    min_child_weight=5,
    random_state=42,
    n_jobs=-1
)
```

```python
final_xgb.fit(X_train_scaled, y_train)
```

```python
y_pred = final_xgb.predict(X_test_scaled)
```

```python
rmse = mean_squared_error(y_test, y_pred)**0.5
r2 = r2_score(y_test, y_pred)

print("Final XGBoost RMSE:", rmse)
print("Final XGBoost R2 Score:", r2)
```
